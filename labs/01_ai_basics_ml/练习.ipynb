# 实验四：强化学习——Q-learning 一维寻宝

环境中有六个位置，智能体从 0 出发，到达 5 获得宝藏：

```text
起点                         宝藏
 0  →  1  →  2  →  3  →  4  →  5
```

智能体每次可以向左或向右：

- 到达终点：奖励 `+10`；
- 普通移动：奖励 `-0.1`；
- 目标：学习一条尽快到达终点的策略。

## 运行前先回答

1. 这里的状态、动作、奖励分别是什么？
2. 为什么普通移动要给予一个很小的负奖励？
3. 如果智能体完全不探索，可能出现什么问题？

````python
# 环境和参数：直接运行
rl_rng = np.random.default_rng(42)

N_STATES = 6
GOAL_STATE = 5
LEFT = 0
RIGHT = 1
ACTION_NAMES = ["LEFT", "RIGHT"]

LEARNING_RATE = 0.1
GAMMA = 0.95
EPSILON_START = 1.0
EPSILON_MIN = 0.05
EPSILON_DECAY = 0.99
EPISODES = 500
MAX_STEPS = 30

print("状态数量：", N_STATES)
print("可选动作：", ACTION_NAMES)
print("终点状态：", GOAL_STATE)
print("✅ 强化学习环境准备完成")
````

## 任务 4：补全 Q-learning 的核心更新量

Q-learning 更新公式是：

$$Q(s,a) ← Q(s,a) + α[target - Q(s,a)]$$

下面已经提供环境交互和探索逻辑。你只需要把数学公式翻译成一行 Python，计算 `update_amount`。

提示：需要使用 `LEARNING_RATE`、`target` 和 `q_table[state, action]`。

````python
def train_q_learning():
    q_table = np.zeros((N_STATES, 2))
    epsilon = EPSILON_START
    rewards = []
    successes = []

    for episode in range(EPISODES):
        state = 0
        total_reward = 0.0
        success = 0

        for step in range(MAX_STEPS):
            # epsilon-greedy：一部分时间探索，一部分时间利用已有经验
            if rl_rng.random() < epsilon:
                action = int(rl_rng.integers(2))
            else:
                action = int(np.argmax(q_table[state]))

            if action == LEFT:
                next_state = max(0, state - 1)
            else:
                next_state = min(GOAL_STATE, state + 1)

            done = next_state == GOAL_STATE
            reward = 10.0 if done else -0.1

            if done:
                target = reward
            else:
                target = reward + GAMMA * np.max(q_table[next_state])

            # TODO：把 None 替换为 Q-learning 的更新量
            update_amount = None

            if update_amount is None:
                return None, None, None

            q_table[state, action] += update_amount
            state = next_state
            total_reward += reward

            if done:
                success = 1
                break

        rewards.append(total_reward)
        successes.append(success)
        epsilon = max(EPSILON_MIN, epsilon * EPSILON_DECAY)

    return q_table, rewards, successes


rl_q_table, rl_rewards, rl_successes = train_q_learning()

if rl_q_table is None:
    print("📝 尚未完成：请根据公式填写 update_amount。")
else:
    print("✅ Q-learning 训练完成")
````

````python
# 自动检查、策略演示与训练曲线
if rl_q_table is None:
    print("⏸️ 请先完成任务 4。")
else:
    print("========== 学到的 Q 表 ==========")
    print("状态\t向左Q值\t向右Q值\t最佳动作")

    for state in range(GOAL_STATE):
        best_action = int(np.argmax(rl_q_table[state]))
        print(
            f"{state}\t{rl_q_table[state, LEFT]:.3f}\t\t"
            f"{rl_q_table[state, RIGHT]:.3f}\t\t{ACTION_NAMES[best_action]}"
        )

    state = 0
    learned_path = [state]

    for _ in range(20):
        action = int(np.argmax(rl_q_table[state]))
        if action == LEFT:
            state = max(0, state - 1)
        else:
            state = min(GOAL_STATE, state + 1)
        learned_path.append(state)
        if state == GOAL_STATE:
            break

    print("\n学到的路径：", " → ".join(map(str, learned_path)))

    if learned_path[-1] == GOAL_STATE:
        print("✅ 成功到达宝藏")
    else:
        print("❌ 尚未学会到达宝藏，请检查更新公式")

    window = 25
    avg_rewards = np.convolve(
        rl_rewards, np.ones(window) / window, mode="valid"
    )
    success_rates = np.convolve(
        rl_successes, np.ones(window) / window, mode="valid"
    )

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(avg_rewards, color="blue")
    axes[0].set_title("Moving Average Reward")
    axes[0].set_xlabel("Episode")
    axes[0].set_ylabel("Average Reward")
    axes[0].grid(alpha=0.3)

    axes[1].plot(success_rates, color="green")
    axes[1].set_title("Success Rate")
    axes[1].set_xlabel("Episode")
    axes[1].set_ylabel("Success Rate")
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.show()
````
